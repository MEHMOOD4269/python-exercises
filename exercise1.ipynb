```python
# Q1. Reverse string and every 2nd character
def string_ops(s):
    return s[::-1], s[::2]

# Q2. Squares of even numbers using list comprehension
def square_evens(nums):
    return [x**2 for x in nums if x % 2 == 0]

# Q3. Frequency of each character using a dictionary
def char_freq(s):
    freq = {}
    for char in s:
        freq[char] = freq.get(char, 0) + 1
    return freq

# Q4. Indices of all odd numbers using enumerate()
def odd_indices(lst):
    return [i for i, val in enumerate(lst) if val % 2 != 0]

# Q5. Rotate a list right by k positions using slicing
def rotate_right(lst, k):
    if not lst: return []
    k = k % len(lst)
    return lst[-k:] + lst[:-k]

# Q6. Two lists to a dictionary using zip()
def lists_to_dict(keys, values):
    return dict(zip(keys, values))

# Q7. iter(), next(), and StopIteration exception handling
def iterate_list(lst):
    my_iter = iter(lst)
    while True:
        try:
            print(next(my_iter))
        except StopIteration:
            print("Iteration finished.")
            break

# Q8. Flatten a nested list using list comprehension
def flatten_list(nested_list):
    return [item for sublist in nested_list for item in sublist]

# Q9. Check palindrome using slicing
def is_palindrome(s):
    return s == s[::-1]

# Q10. Second largest number in a list (without sorting)
def second_largest(lst):
    unique_lst = list(set(lst)) 
    if len(unique_lst) < 2: return None
    unique_lst.remove(max(unique_lst))
    return max(unique_lst)
# This Was Done By Mehmood Nasir
```
